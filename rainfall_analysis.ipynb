Clean machakos_raw using a function with try/except that converts valid values to integers. Replace invalid values with the average of the valid Machakos values (rounded), and print which months were imputed.

```python
import numpy as np

# County names
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

# Month names
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

# Rainfall data for Nairobi, Nakuru, Kisumu and Mombasa
clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]
]

# Raw Machakos rainfall data
machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]

```

Clean machakos_raw using a function with try/except that converts valid values to integers. Replace invalid values with the average of the valid Machakos values (rounded), and print which months were imputed.

```python
#Identify the valid readings and calculate the average
valid_values = []

for value in machakos_raw:
    try:
        number = int(value.strip())
        valid_values.append(number)

    except ValueError:
        pass

average_rainfall = round(sum(valid_values) / len(valid_values))

print("Valid readings:", valid_values)
print("Number of valid readings:", len(valid_values))
print("Average rainfall:", average_rainfall, "mm")


#Create the cleaning function
def clean_machakos(data):
    cleaned_data = []
    imputed_months = []

    for index, value in enumerate(data):
        try:
            number = int(value.strip())
            cleaned_data.append(number)

        except ValueError:
            cleaned_data.append(average_rainfall)
            imputed_months.append(months[index])

    print("Months imputed:", imputed_months)

    return cleaned_data


machakos_clean = clean_machakos(machakos_raw)

print("Cleaned Machakos data:", machakos_clean)


```

Combine all five counties into a single 5 × 12 NumPy array and confirm its shape and dtype.

```python
# Combine the four clean rows with Machakos data
all_rows = clean_rows + [machakos_clean]

# Convert the combined data into a NumPy array
R = np.array(all_rows, dtype=int)

print("Rainfall array:")
print(R)

print("Array shape:", R.shape)
print("Array data type:", R.dtype)
```

Calculate annual total rainfall per county and the average rainfall per month across all counties.

```python
#Calculate annual total rainfall per county
annual_totals = []

for row in R:
    total = np.sum(row)
    annual_totals.append(total)

for county, total in zip(counties, annual_totals):
    print(county, ":", total, "mm")

#Calculate average rainfall per month across all counties
monthly_averages = []

for column in zip(*R):
    average = np.mean(column)
    monthly_averages.append(round(average, 2))

for month, average in zip(months, monthly_averages):
    print(month, ":", average, "mm")
```

Using np.argmax with axis=1, find the wettest month for each county and print it by month name (use zip() from Section 2.9 to pair counties with results).

```python
# Find the index of the highest rainfall for each county
wettest_indices = np.argmax(R, axis=1)

# Pair counties with their wettest month indices
for county, index in zip(counties, wettest_indices):
    print(county, ":", months[index])
```

Count, for each county, the number of “dry months” (below 50 mm) using a boolean mask.

```python
# Create a Boolean mask for rainfall below 50 mm
dry_month_mask = R < 50

print("Boolean mask:")
print(dry_month_mask)

# Count the dry months for each county
dry_month_counts = np.sum(dry_month_mask, axis=1)

# Display results
for county, count in zip(counties, dry_month_counts):
    print(county, ":", count, "dry months")
```

Identify the two consecutive months with the highest combined rainfall for each county; this is the recommended planting window. (Hint: the shifted-array technique R[:, :-1] + R[:, 1:] in Section 2.10.)

```python
# Calculate rainfall for every consecutive two-month pair
pair_totals = R[:, :-1] + R[:, 1:]

# Find the index of the highest combined rainfall
best_pair_indices = np.argmax(pair_totals, axis=1)

# Display the recommended planting windows
for county, index in zip(counties, best_pair_indices):
    planting_window = months[index] + "-" + months[index + 1]

    print(county, ":", planting_window)
```

Write a CSV file rainfall_summary.csv with columns county, annual_total_mm, wettest_month, dry_months and planting_window

```python
import csv

# Define the required CSV columns
import os

fieldnames = [
    "county",
    "annual_total_mm",
    "wettest_month",
    "dry_months",
    "planting_window"
]

# Create the CSV file
with open("rainfall_summary.csv", "w", newline="") as file:

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    # Write the column headings
    writer.writeheader()

    # Write one row for each county
    for county, total, wet_index, dry_count, pair_index in zip(
        counties,
        annual_totals,
        wettest_indices,
        dry_month_counts,
        best_pair_indices
    ):

        writer.writerow({
            "county": county,
            "annual_total_mm": int(total),
            "wettest_month": months[wet_index],
            "dry_months": int(dry_count),
            "planting_window": months[pair_index] + "-" + months[pair_index + 1]
        })

print("rainfall_summary.csv created successfully!")
```
